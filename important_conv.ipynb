```python
import pandas as pd
from io import StringIO

# ============================================================
# PANDAS COMPLETE NOTES - TABULAR OUTPUT
# ============================================================

# Sample DataFrame
data = {
    'Name': ['Piyush', 'Rahul', 'Aman', 'Priya', 'Neha'],
    'Age': [20, 21, 19, 22, 20],
    'City': ['Jodhpur', 'Jaipur', 'Jodhpur', 'Kota', 'Jaipur'],
    'Marks': [85, 78, 92, 88, 75],
    'Salary': [25000, 30000, 28000, 35000, 27000]
}

df = pd.DataFrame(data)

print("\n========== ORIGINAL DATAFRAME ==========")
print(df)


# ============================================================
# 1. CREATE SERIES
# ============================================================

series = pd.Series([10, 20, 30, 40, 50])

print("\n========== SERIES ==========")
print(series)


# ============================================================
# 2. BASIC INFORMATION
# ============================================================

basic_info = pd.DataFrame({
    'Operation': [
        'Shape',
        'Columns',
        'Data Types',
        'Number of Rows',
        'Number of Columns'
    ],
    'Output': [
        str(df.shape),
        str(list(df.columns)),
        str(df.dtypes.to_dict()),
        len(df),
        len(df.columns)
    ]
})

print("\n========== BASIC INFORMATION ==========")
print(basic_info.to_string(index=False))


# ============================================================
# 3. HEAD AND TAIL
# ============================================================

print("\n========== HEAD() ==========")
print(df.head())

print("\n========== TAIL() ==========")
print(df.tail())


# ============================================================
# 4. STATISTICAL FUNCTIONS
# ============================================================

statistics = pd.DataFrame({
    'Function': ['Mean', 'Median', 'Maximum', 'Minimum', 'Sum', 'Standard Deviation'],
    'Marks': [
        df['Marks'].mean(),
        df['Marks'].median(),
        df['Marks'].max(),
        df['Marks'].min(),
        df['Marks'].sum(),
        df['Marks'].std()
    ]
})

print("\n========== STATISTICAL FUNCTIONS ==========")
print(statistics.to_string(index=False))


# ============================================================
# 5. SELECTING COLUMNS
# ============================================================

print("\n========== SINGLE COLUMN ==========")
print(df['Name'])

print("\n========== MULTIPLE COLUMNS ==========")
print(df[['Name', 'Age', 'Marks']])


# ============================================================
# 6. LOC AND ILOC
# ============================================================

loc_output = pd.DataFrame({
    'Method': ['loc', 'iloc'],
    'Code': [
        "df.loc[0, 'Name']",
        "df.iloc[0, 1]"
    ],
    'Output': [
        df.loc[0, 'Name'],
        df.iloc[0, 1]
    ]
})

print("\n========== LOC AND ILOC ==========")
print(loc_output.to_string(index=False))


# ============================================================
# 7. FILTERING
# ============================================================

print("\n========== MARKS > 80 ==========")
print(df[df['Marks'] > 80])

print("\n========== AGE > 19 ==========")
print(df[df['Age'] > 19])

print("\n========== MULTIPLE CONDITIONS ==========")
print(df[(df['Age'] > 19) & (df['Marks'] > 80)])


# ============================================================
# 8. ADDING NEW COLUMN
# ============================================================

df['Grade'] = ['A', 'B', 'A+', 'A', 'B']

print("\n========== ADD NEW COLUMN ==========")
print(df)


# ============================================================
# 9. RENAMING COLUMN
# ============================================================

rename_df = df.rename(columns={'Marks': 'Score'})

print("\n========== RENAME COLUMN ==========")
print(rename_df)


# ============================================================
# 10. SORTING
# ============================================================

print("\n========== SORT ASCENDING ==========")
print(df.sort_values('Marks'))

print("\n========== SORT DESCENDING ==========")
print(df.sort_values('Marks', ascending=False))


# ============================================================
# 11. VALUE COUNTS
# ============================================================

print("\n========== VALUE COUNTS ==========")
print(df['City'].value_counts())


# ============================================================
# 12. GROUPBY
# ============================================================

grouped = df.groupby('City')['Marks'].mean().reset_index()

print("\n========== GROUPBY ==========")
print(grouped.to_string(index=False))


# ============================================================
# 13. MISSING VALUES
# ============================================================

df_missing = df.copy()

df_missing.loc[2, 'Marks'] = None
df_missing.loc[4, 'City'] = None

print("\n========== DATA WITH MISSING VALUES ==========")
print(df_missing)

print("\n========== ISNULL() ==========")
print(df_missing.isnull())

print("\n========== MISSING VALUE COUNT ==========")
print(df_missing.isnull().sum())


# ============================================================
# 14. FILL MISSING VALUES
# ============================================================

filled_df = df_missing.copy()
filled_df['Marks'] = filled_df['Marks'].fillna(0)
filled_df['City'] = filled_df['City'].fillna('Unknown')

print("\n========== FILLNA() ==========")
print(filled_df)


# ============================================================
# 15. DROP MISSING VALUES
# ============================================================

drop_df = df_missing.dropna()

print("\n========== DROPNA() ==========")
print(drop_df)


# ============================================================
# 16. DUPLICATES
# ============================================================

duplicate_df = pd.DataFrame({
    'Name': ['Aman', 'Rahul', 'Aman', 'Priya'],
    'Marks': [80, 75, 80, 90]
})

print("\n========== DUPLICATE DATA ==========")
print(duplicate_df)

print("\n========== DUPLICATED() ==========")
print(duplicate_df.duplicated())

print("\n========== DROP_DUPLICATES() ==========")
print(duplicate_df.drop_duplicates())


# ============================================================
# 17. DELETE COLUMN
# ============================================================

delete_column_df = df.copy()
delete_column_df.drop('Salary', axis=1, inplace=True)

print("\n========== DROP COLUMN ==========")
print(delete_column_df)


# ============================================================
# 18. DELETE ROW
# ============================================================

delete_row_df = df.copy()
delete_row_df.drop(0, axis=0, inplace=True)

print("\n========== DROP ROW ==========")
print(delete_row_df)


# ============================================================
# 19. CONCAT
# ============================================================

df1 = pd.DataFrame({
    'Name': ['A', 'B'],
    'Marks': [80, 90]
})

df2 = pd.DataFrame({
    'Name': ['C', 'D'],
    'Marks': [75, 85]
})

concat_df = pd.concat([df1, df2], ignore_index=True)

print("\n========== CONCAT ==========")
print(concat_df)


# ============================================================
# 20. MERGE
# ============================================================

students = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Aman', 'Rahul', 'Priya']
})

marks = pd.DataFrame({
    'ID': [1, 2, 3],
    'Marks': [85, 90, 78]
})

merged_df = pd.merge(students, marks, on='ID', how='inner')

print("\n========== MERGE ==========")
print(merged_df)


# ============================================================
# 21. DESCRIBE
# ============================================================

print("\n========== DESCRIBE() ==========")
print(df.describe())


# ============================================================
# 22. INFO
# ============================================================

print("\n========== INFO() ==========")
df.info()


# ============================================================
# 23. CREATE DATAFRAME FROM CSV STRING
# ============================================================

csv_data = """Name,Age,Marks
Aman,20,85
Rahul,21,90
Priya,19,88"""

csv_df = pd.read_csv(StringIO(csv_data))

print("\n========== READ CSV ==========")
print(csv_df)


# ============================================================
# 24. SAVE DATA
# ============================================================

# Uncomment these when you want to create files

# df.to_csv('output.csv', index=False)
# df.to_excel('output.xlsx', index=False)
# df.to_json('output.json')
# df.to_html('output.html')


# ============================================================
# FINAL COMMAND SUMMARY TABLE
# ============================================================

commands = pd.DataFrame({
    'Operation': [
        'Import Pandas',
        'Create Series',
        'Create DataFrame',
        'Read CSV',
        'Read Excel',
        'Read JSON',
        'Read Text',
        'Read TSV',
        'Read HTML',
        'Read SQL',
        'Read Parquet',
        'Read Feather',
        'First Rows',
        'Last Rows',
        'Shape',
        'Columns',
        'Data Types',
        'Information',
        'Statistics',
        'Select Column',
        'Select Rows',
        'Filter Data',
        'Add Column',
        'Rename Column',
        'Delete Column',
        'Delete Row',
        'Missing Values',
        'Fill Missing Values',
        'Remove Missing Rows',
        'Remove Duplicates',
        'Sort Data',
        'Value Counts',
        'GroupBy',
        'Merge',
        'Concat',
        'Save CSV',
        'Save Excel',
        'Save JSON'
    ],

    'Code': [
        'import pandas as pd',
        'pd.Series([1,2,3])',
        'pd.DataFrame(data)',
        "pd.read_csv('file.csv')",
        "pd.read_excel('file.xlsx')",
        "pd.read_json('file.json')",
        "pd.read_csv('file.txt', delimiter=' ')",
        "pd.read_csv('file.tsv', sep='\\t')",
        "pd.read_html('file.html')",
        "pd.read_sql(query, connection)",
        "pd.read_parquet('file.parquet')",
        "pd.read_feather('file.feather')",
        'df.head()',
        'df.tail()',
        'df.shape',
        'df.columns',
        'df.dtypes',
        'df.info()',
        'df.describe()',
        "df['Name']",
        'df.iloc[0:5]',
        "df[df['Age'] > 18]",
        "df['Salary'] = values",
        "df.rename(columns={'Age':'Years'})",
        "df.drop('Age', axis=1)",
        "df.drop(0, axis=0)",
        'df.isnull()',
        'df.fillna(0)',
        'df.dropna()',
        'df.drop_duplicates()',
        "df.sort_values('Age')",
        "df['City'].value_counts()",
        "df.groupby('City')['Salary'].mean()",
        "pd.merge(df1, df2, on='ID')",
        'pd.concat([df1, df2])',
        "df.to_csv('output.csv', index=False)",
        "df.to_excel('output.xlsx', index=False)",
        "df.to_json('output.json')"
    ]
})

print("\n\n============================================================")
print("              PANDAS COMMAND CHEAT SHEET")
print("============================================================")

print(commands.to_string(index=False))
```

| **Format**                        | **Extension**          | **Pandas Method to Load**                |
| --------------------------------- | ---------------------- | ---------------------------------------- |
| CSV (Comma-Separated Values)      | `.csv`                 | `pd.read_csv('file.csv')`                |
| Excel                             | `.xls`, `.xlsx`        | `pd.read_excel('file.xlsx')`             |
| JSON (JavaScript Object Notation) | `.json`                | `pd.read_json('file.json')`              |
| Text                              | `.txt`                 | `pd.read_csv('file.txt', delimiter=' ')` |
| TSV (Tab-Separated Values)        | `.tsv`                 | `pd.read_csv('file.tsv', sep='\t')`      |
| HTML Tables                       | `.html`                | `pd.read_html('file.html')`              |
| SQL Database                      | `.db`, `.sqlite`, etc. | `pd.read_sql(query, connection)`         |
| Parquet                           | `.parquet`             | `pd.read_parquet('file.parquet')`        |
| Feather                           | `.feather`             | `pd.read_feather('file.feather')`        |


| **Format / Source**             | **Extension / Type** | **Pandas Method / Code**                                         |
| ------------------------------- | -------------------- | ---------------------------------------------------------------- |
| HDF5 (Hierarchical Data Format) | `.h5`, `.hdf5`       | `pd.read_hdf('file.h5')`                                         |
| ORC (Optimized Row Columnar)    | `.orc`               | `pd.read_orc('file.orc')` *(Pandas 1.5+)*                        |
| ORC via PyArrow                 | `.orc`               | `pyarrow.orc.ORCFile('file.orc')`                                |
| URL / Web Data                  | URL                  | `pd.read_csv('https://example.com/file.csv')`                    |
| Google Sheets using gspread     | Google Sheet         | `gspread` + `pd.DataFrame()`                                     |
| Google Sheets using API         | Google Sheets API    | `pd.read_csv(sheet_url)` or Google Sheets API + `pd.DataFrame()` |


| **Method**           | **Code**                                  |
| -------------------- | ----------------------------------------- |
| Using `gspread`      | `import gspread`                          |
| Open Google Sheet    | `sheet = client.open("SheetName").sheet1` |
| Get all records      | `data = sheet.get_all_records()`          |
| Convert to DataFrame | `df = pd.DataFrame(data)`                 |
