```python
import numpy as np
import pandas as pd
```

```python
Employees = pd.DataFrame({
    'EMPLOYEE_ID' : [1,2,3,4,5],
    'NAME' : ['Ram', 'Shyam', 'Dheeraj', 'Piyush', 'Ajay'],
    'AGE' : [22,32,21,28,32],
    'DEPARTMENT' : ['HR', 'IT', 'FINANCE', 'MARKETING', 'PA'],
})

# DATA FRAME 2 SALARY INFORMATION.
salaries = pd.DataFrame({
    'EMPLOYEE_ID' : [1,2,3,6,7],
     'SALARY' : [60000,80000,45000,56000,90000],
     'BONUS' : [5000,3100,11000,15000,8000]
    
})
```

```python
Employees
```

```python
salaries
```

```python
pd.merge(Employees,salaries,on='EMPLOYEE_ID' , how='inner')
```

```python
pd.merge(Employees,salaries,on='EMPLOYEE_ID' , how='outer')
```

```python
pd.merge(Employees,salaries,on='EMPLOYEE_ID' , how='left')
```

```python
pd.merge(Employees,salaries,on='EMPLOYEE_ID' , how='right')
```

## concartination

```python
pd.concat([Employees,salaries])
```

```python
pd.concat([Employees,salaries], axis=1)
```

# joining two data frames

```python
df1 = pd. DataFrame ( { 
    'name': [ 'Alice', 'Bob', 'Charlie' ]
  }
    , index=[1, 2, 3])
# second data frames
df2 = pd.DataFrame({
    'Score' : [85,98,78]
}, index = [2,3,4])
```

```python
df1
```

```python
df2
```

```python
df1.join(df2, how='outer')
```

```python
df2.join(df1, how='outer')
```

```python
df1.join(df2, how='inner')
```
