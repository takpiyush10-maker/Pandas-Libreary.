```python
import pandas as pd
import numpy as np
```

```python
Data = {
    'A' : [1,2,np.nan,4,5],
    'B' : [np.nan,2,3,4,5],
    'C' : [1,2,3,np.nan,np.nan],
    'D' : [1,np.nan,np.nan,np.nan,5]
}
df = pd.DataFrame(Data)
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

```python
## removing missing value
df.dropna()
```

```python
df.dropna(thresh=3)
# Rows me non null values kitni honi chaiyea atleast jo ham likh rahe hai
```

```python
## Filing the missing data
df.fillna(0)
```

```python
values = {'A':0,'B':100,'C':300,'D':400}
df.fillna(value=values)
```

```python
df.fillna(df.mean())
```
